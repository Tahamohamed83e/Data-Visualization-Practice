Clipping

```python
Q1= df['Fare'].quantile(0.25)
Q3= df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
lower
upper
```

```python
df['Fare'].describe()
```

```python
df['Fare']=df['Fare'].clip(lower=lower , upper=upper)
df = df[(df['Fare'] >= lower) & (df['Fare'] <= upper)]
df['Fare'].describe()
```
